### Dashboard da Feira Online (Streamlit)
O código fornecido utiliza a biblioteca `streamlit` para criar uma interface web. Como o Streamlit não roda diretamente dentro das células do Jupyter (ele exige que o script seja rodado por fora via terminal), adicionei o comando mágico `%%writefile` na célula abaixo.

Isso fará com que, ao rodar a célula, todo o código seja salvo em um arquivo chamado `app_feira.py` na mesma pasta do seu Jupyter Notebook.

```python
%%writefile app_feira.py
import streamlit as st
import mysql.connector
import pandas as pd

st.set_page_config(page_title="Feira Online Sustentável - RN", layout="wide")

# Configuração de conexão com o MySQL
def get_connection(database_name):
    return mysql.connector.connect(
        host="localhost",
        user="root",          # Altere para seu usuário do MySQL
        password="sua_senha", # Altere para sua senha do MySQL
        database=database_name
    )

def query_df(sql, db="db_feira_online", params=None):
    conn = get_connection(db)
    df = pd.read_sql(sql, conn, params=params)
    conn.close()
    return df

def execute_sql(sql, params=None, db="db_feira_online"):
    conn = get_connection(db)
    cursor = conn.cursor()
    cursor.execute(sql, params)
    conn.commit()
    cursor.close()
    conn.close()

st.title("🌱 Feira Online Sustentável — Gestão & Inteligência de Dados")
st.markdown("Integração **MySQL + Python (Streamlit)** | Dados Públicos (MAPA/CONAB) e CRUD Operacional")

aba_catalogo, aba_crud_prod, aba_crud_produtor, aba_analise = st.tabs([
    "🛒 1. Catálogo e Consultas (READ)",
    "➕✏️🗑️ 2. Gerenciar Produtos (INSERT / UPDATE / DELETE)",
    "👩‍🌾 3. Gerenciar Produtores",
    "📊 4. Análise de Dados Públicos (CNPO & CONAB)"
])

# ==============================================================================
# ABA 1: CONSULTA (SELECT COM FILTROS E VIEWS)
# ==============================================================================
with aba_catalogo:
    st.subheader("Catálogo de Alimentos Frescos (View `vw_catalogo_feira`)")
    df_cat = query_df("SELECT * FROM vw_catalogo_feira")
    
    col_f1, col_f2 = st.columns(2)
    with col_f1:
        cidades = ["Todas"] + sorted(df_cat["cidade_origem"].unique().tolist())
        filtro_cidade = st.selectbox("Filtrar por Município de Origem:", cidades)
    with col_f2:
        categorias = ["Todas"] + sorted(df_cat["nome_categoria"].unique().tolist())
        filtro_cat = st.selectbox("Filtrar por Categoria:", categorias)

    if filtro_cidade != "Todas":
        df_cat = df_cat[df_cat["cidade_origem"] == filtro_cidade]
    if filtro_cat != "Todas":
        df_cat = df_cat[df_cat["nome_categoria"] == filtro_cat]

    st.dataframe(df_cat, use_container_width=True)

    st.subheader("Desempenho de Vendas por Produtor (View `vw_faturamento_produtor`)")
    df_fat = query_df("SELECT * FROM vw_faturamento_produtor ORDER BY receita_total_reais DESC")
    st.dataframe(df_fat, use_container_width=True)

# ==============================================================================
# ABA 2: INSERÇÃO, ALTERAÇÃO E REMOÇÃO DE PRODUTOS (CRUD COMPLETO)
# ==============================================================================
with aba_crud_prod:
    col_ins, col_upd, col_del = st.columns(3)

    produtores_df = query_df("SELECT id_produtor, nome_produtor, cidade FROM tb_produtor WHERE situacao='ATIVO'")
    categorias_df = query_df("SELECT id_categoria, nome_categoria FROM tb_categoria")
    produtos_df = query_df("SELECT id_produto, nome_produto, preco_unitario, estoque_disponivel FROM tb_produto")

    # --- OPERAÇÃO CREATE (INSERT) ---
    with col_ins:
        st.markdown("### ➕ Cadastrar Produto")
        with st.form("form_insert_produto"):
            prod_opcoes = {f"{r['nome_produtor']} ({r['cidade']})": r["id_produtor"] for _, r in produtores_df.iterrows()}
            cat_opcoes = {r["nome_categoria"]: r["id_categoria"] for _, r in categorias_df.iterrows()}
            
            produtor_sel = st.selectbox("Produtor", list(prod_opcoes.keys()))
            categoria_sel = st.selectbox("Categoria", list(cat_opcoes.keys()))
            nome_prod = st.text_input("Nome do Produto")
            unidade = st.selectbox("Unidade", ["kg", "unidade", "maco", "bandeja", "duzia", "litro"])
            preco = st.number_input("Preço Unitário (R$)", min_value=0.50, value=5.00, step=0.50)
            estoque = st.number_input("Estoque Inicial", min_value=0.0, value=20.0, step=1.0)
            organico = st.checkbox("Orgânico Certificado (CNPO)", value=True)
            
            if st.form_submit_button("Inserir no MySQL"):
                sql_ins = """
                    INSERT INTO tb_produto 
                    (id_produtor, id_categoria, nome_produto, unidade_medida, preco_unitario, estoque_disponivel, organico_certificado)
                    VALUES (%s, %s, %s, %s, %s, %s, %s)
                """
                execute_sql(sql_ins, (prod_opcoes[produtor_sel], cat_opcoes[categoria_sel], nome_prod, unidade, preco, estoque, organico))
                st.success(f"Produto '{nome_prod}' cadastrado com sucesso!")
                st.rerun()

    # --- OPERAÇÃO UPDATE (ALTERAÇÃO) ---
    with col_upd:
        st.markdown("### ✏️ Atualizar Preço / Estoque")
        with st.form("form_update_produto"):
            mapa_prod = {f"#{r['id_produto']} - {r['nome_produto']}": r for _, r in produtos_df.iterrows()}
            prod_chave = st.selectbox("Selecione o Produto", list(mapa_prod.keys()))
            registro_atual = mapa_prod[prod_chave]
            
            novo_preco = st.number_input("Novo Preço (R$)", min_value=0.50, value=float(registro_atual["preco_unitario"]), step=0.50)
            novo_estoque = st.number_input("Novo Estoque", min_value=0.0, value=float(registro_atual["estoque_disponivel"]), step=1.0)
            
            if st.form_submit_button("Atualizar no MySQL"):
                sql_upd = "UPDATE tb_produto SET preco_unitario = %s, estoque_disponivel = %s WHERE id_produto = %s"
                execute_sql(sql_upd, (novo_preco, novo_estoque, int(registro_atual["id_produto"])))
                st.info("Registro atualizado com sucesso!")
                st.rerun()

    # --- OPERAÇÃO DELETE (REMOÇÃO) ---
    with col_del:
        st.markdown("### 🗑️ Remover Produto")
        with st.form("form_delete_produto"):
            prod_del_chave = st.selectbox("Produto a Remover", list(mapa_prod.keys()))
            id_del = int(mapa_prod[prod_del_chave]["id_produto"])
            
            if st.form_submit_button("Excluir do MySQL"):
                try:
                    execute_sql("DELETE FROM tb_produto WHERE id_produto = %s", (id_del,))
                    st.warning(f"Produto #{id_del} removido do banco!")
                    st.rerun()
                except Exception as e:
                    st.error(f"Não é possível remover produto vinculado a pedidos: {e}")

# ==============================================================================
# ABA 3: CADASTRO DE PRODUTORES (ALINHADO AO CNPO)
# ==============================================================================
with aba_crud_produtor:
    st.subheader("Cadastrar Novo Pequeno Produtor / Agricultor Familiar")
    with st.form("form_novo_produtor"):
        c1, c2, c3 = st.columns(3)
        nome_p = c1.text_input("Nome do Produtor")
        doc_p = c2.text_input("CPF ou CNPJ")
        cert_p = c3.selectbox("Modalidade", ["OCS", "OPAC", "CERTIFICADORA", "AGRICULTURA_FAMILIAR"])
        ent_p = c1.text_input("Entidade / Associação (OCS/OPAC)")
        cid_p = c2.text_input("Município", value="NATAL")
        uf_p = c3.text_input("UF", value="RN", max_chars=2)
        
        if st.form_submit_button("Salvar Produtor"):
            sql_p = """
                INSERT INTO tb_produtor (nome_produtor, cpf_cnpj, tipo_certificacao, entidade_certificadora, cidade, uf)
                VALUES (%s, %s, %s, %s, %s, %s)
            """
            execute_sql(sql_p, (nome_p, doc_p, cert_p, ent_p, cid_p.upper(), uf_p.upper()))
            st.success(f"Produtor {nome_p} inserido no MySQL!")
            st.rerun()

    st.dataframe(query_df("SELECT * FROM tb_produtor"), use_container_width=True)

# ==============================================================================
# ABA 4: CONSULTAS ESTATÍSTICAS NO BANCO DE DADOS PÚBLICOS (db_analise_publica)
# ==============================================================================
with aba_analise:
    st.subheader("Indicadores Estatísticos (MAX, MIN, SUM, AVG) — `db_analise_publica`")
    sql_vb_rn = """
        SELECT municipio_armazem_venda AS Polo_RN,
               SUM(qtd_produto_kg) AS Total_kg,
               ROUND(AVG(qtd_produto_kg), 2) AS Media_Mensal_kg,
               MIN(qtd_produto_kg) AS Min_kg,
               MAX(qtd_produto_kg) AS Max_kg,
               ROUND(SUM(valor_comercializado), 2) AS Total_Reais,
               SUM(clientes_atendidos) AS Clientes_Atendidos
        FROM conab_venda_balcao
        WHERE uf = 'RN'
        GROUP BY municipio_armazem_venda
        ORDER BY Total_kg DESC
    """
    df_rn = query_df(sql_vb_rn, db="db_analise_publica")
    st.dataframe(df_rn, use_container_width=True)
    st.bar_chart(df_rn.set_index("Polo_RN")["Total_kg"])
```
